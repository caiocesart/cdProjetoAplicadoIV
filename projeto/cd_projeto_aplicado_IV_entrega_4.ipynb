# **Referencial Teórico**

A previsão de carga elétrica é um problema clássico de séries temporais, amplamente estudado devido à sua importância para planejamento energético, gerenciamento de redes elétricas e otimização de recursos. Segundo Hyndman e Athanasopoulos (2021), modelos de previsão temporal buscam identificar padrões históricos para estimar valores futuros, considerando componentes como tendência, sazonalidade e variabilidade.

Entre os métodos estatísticos tradicionais, destacam-se os modelos ARIMA (Autoregressive Integrated Moving Average) e SARIMA (Seasonal ARIMA). Esses modelos são amplamente utilizados por sua capacidade de representar dependências temporais lineares e componentes sazonais (BOX et al., 2015). O modelo SARIMA, especificamente, adiciona parâmetros sazonais ao ARIMA convencional, permitindo modelar padrões periódicos recorrentes, como sazonalidade semanal ou anual no consumo elétrico.

Apesar de sua interpretabilidade, modelos estatísticos apresentam limitações na modelagem de relações não lineares e padrões complexos. Em razão disso, técnicas de aprendizado de máquina passaram a ser aplicadas em problemas de previsão energética. Métodos como Random Forest e Gradient Boosting conseguem capturar interações complexas entre variáveis temporais e atributos derivados da série, frequentemente apresentando desempenho superior em cenários com alta variabilidade (HONG; FAN, 2016).

Outra abordagem amplamente estudada envolve modelos de deep learning, especialmente redes neurais recorrentes (RNN) e Long Short-Term Memory (LSTM). Segundo Hochreiter e Schmidhuber (1997), as redes LSTM conseguem aprender dependências de longo prazo em sequências temporais, sendo adequadas para problemas com forte dependência histórica. Entretanto, tais modelos exigem maior volume de dados, maior capacidade computacional e possuem menor interpretabilidade quando comparados a modelos estatísticos tradicionais.

Além das técnicas de modelagem, conceitos fundamentais de séries temporais são essenciais para compreensão do problema. A estacionariedade, por exemplo, refere-se à estabilidade estatística da série ao longo do tempo e influencia diretamente o desempenho de modelos autorregressivos. Já a autocorrelação mede o grau de dependência entre observações em diferentes instantes temporais, sendo frequentemente analisada por meio dos gráficos ACF e PACF.

No presente projeto, optou-se inicialmente pela utilização de modelos baseline e do modelo SARIMA, considerando sua capacidade de modelar sazonalidade semanal observada nos dados explorados. Como diferencial metodológico, foram aplicadas técnicas de engenharia de atributos, incluindo criação de lags e médias móveis, visando capturar dependências temporais relevantes para futuras etapas de aprendizado de máquina.

# **EDA e Pré-processamento dos dados**

A base utilizada neste projeto é a **Electricity Load Diagrams 2011–2014**, composta por medições de consumo elétrico de 370 clientes, registradas em intervalos de 15 minutos. Cada coluna representa um consumidor e a primeira coluna contém a informação temporal. Como o objetivo do trabalho é realizar previsão de carga elétrica de forma viável e interpretável, foi adotado como recorte principal o cliente **MT_124**, permitindo uma análise mais detalhada do comportamento temporal de um único consumidor.

Inicialmente, os dados foram carregados em formato `.txt`, utilizando separador `;` e vírgula como separador decimal, conforme a estrutura original da base. A coluna temporal foi convertida para índice do tipo `datetime`, permitindo a aplicação de operações específicas de séries temporais, como reamostragem, decomposição sazonal e criação de variáveis defasadas.

Como a base original apresenta medições a cada 15 minutos, foi realizada uma transformação da série para frequência diária. Para isso, os valores originalmente em potência foram convertidos para energia em kWh, dividindo cada registro por 4, pois cada medição representa um quarto de hora. Em seguida, as 96 medições diárias foram somadas, gerando uma série diária de consumo elétrico. Essa compactação reduziu a granularidade dos dados, diminuiu ruídos de curtíssimo prazo e tornou a modelagem mais adequada ao escopo do projeto.

Durante a análise de qualidade dos dados, verificou-se que a base não apresentava valores nulos explícitos. Entretanto, alguns registros zerados poderiam estar relacionados a períodos de entrada posterior de clientes na base ou a ajustes de horário de verão. Por isso, foram inspecionados dias com consumo igual a zero e possíveis valores extremos por meio do intervalo interquartil. Os consumos iguais a zero foram substituídos por valores ausentes e tratados por interpolação temporal, preservando a continuidade da série.

A etapa exploratória buscou identificar padrões de tendência, sazonalidade e dependência temporal. Foram gerados gráficos da série diária, decomposição aditiva, boxplots por dia da semana e por mês, além de gráficos de autocorrelação e autocorrelação parcial. A análise indicou a presença de sazonalidade semanal, com diferenças entre dias úteis e finais de semana, além de padrões sazonais ao longo dos meses. Também foi observada forte dependência temporal, justificando o uso de modelos baseados em defasagens e modelos estatísticos de séries temporais.

Para enriquecer a base de modelagem, foram criadas variáveis temporais e variáveis derivadas da própria série, como dia da semana, mês, dia do ano, lags de 1, 7 e 30 dias, além de médias móveis de 7 e 30 dias. Essas variáveis permitem capturar comportamentos recentes, padrões semanais e tendências suavizadas do consumo. Após a criação dessas variáveis, os registros iniciais com valores ausentes decorrentes dos lags e médias móveis foram removidos.

Por fim, a base foi dividida respeitando a ordem cronológica da série temporal. Os anos de **2011 a 2013** foram utilizados como conjunto de treinamento, enquanto o ano de **2014** foi reservado para teste. Essa escolha evita vazamento de informação e simula uma situação real de previsão, em que dados passados são usados para estimar valores futuros.

A análise exploratória evidenciou algumas limitações importantes. A primeira delas está relacionada ao recorte adotado, pois o estudo foi conduzido apenas com o cliente **MT_124**, o que limita a generalização dos resultados para todos os consumidores da base. Além disso, a agregação dos dados para frequência diária simplifica o problema e reduz ruídos, mas também elimina informações intradiárias, como horários de pico, variações ao longo do dia e possíveis padrões específicos de consumo em determinados períodos.

Outra limitação está na ausência de variáveis externas, como temperatura, feriados, tipo de atividade do consumidor ou informações econômicas. Essas variáveis poderiam contribuir para explicar melhor determinadas oscilações no consumo, mas não estavam disponíveis diretamente na base utilizada. Ainda assim, as variáveis derivadas da própria série temporal permitiram capturar parte relevante da dinâmica dos dados, especialmente os padrões de curto prazo e a sazonalidade semanal.

Dessa forma, a etapa de EDA foi fundamental para compreender a estrutura da série, avaliar a qualidade dos dados, justificar os tratamentos aplicados e orientar a escolha dos modelos propostos. As decisões metodológicas adotadas buscaram equilibrar simplicidade, viabilidade computacional e coerência com o objetivo central do projeto, que é prever o consumo diário de energia elétrica a partir do histórico observado.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (14, 6)

# Leitura da base original
df = pd.read_csv(
    '../data/LD2011_2014.txt',
    sep=';',
    index_col=0,
    parse_dates=True,
    decimal=','
)

df.index.name = 'datetime'
df.columns = df.columns.str.strip()

print(f'Shape original: {df.shape}')
print(f'Período: {df.index.min()} a {df.index.max()}')
print(f'Quantidade de clientes: {len(df.columns)}')
print(f'Valores nulos: {df.isnull().sum().sum()}')

display(df.head())
```

```python
# Seleção do cliente MT_124
# Conversão das medições de 15 minutos para kWh
mt_124_kwh = df['MT_124'] / 4

# Agregação diária por soma
mt_124_daily = mt_124_kwh.resample('D').sum().to_frame(name='consumo_kwh')

print(f'Dias totais: {len(mt_124_daily)}')
print(f'Período: {mt_124_daily.index.min().date()} a {mt_124_daily.index.max().date()}')

display(mt_124_daily.describe())
```

```python
# Verificação de dias com consumo zero e possíveis outliers
dias_zero = mt_124_daily[mt_124_daily['consumo_kwh'] == 0]

q1 = mt_124_daily['consumo_kwh'].quantile(0.25)
q3 = mt_124_daily['consumo_kwh'].quantile(0.75)
iqr = q3 - q1
limite_superior = q3 + 1.5 * iqr

dias_alto = mt_124_daily[mt_124_daily['consumo_kwh'] > limite_superior]

print(f'Dias com consumo zero: {len(dias_zero)}')
print(f'Limite superior pelo IQR: {limite_superior:.2f} kWh')
print(f'Dias acima do limite: {len(dias_alto)}')
```

```python
# Tratamento de valores zerados por interpolação temporal
mt_124_daily.loc[mt_124_daily['consumo_kwh'] == 0, 'consumo_kwh'] = np.nan
mt_124_daily['consumo_kwh'] = mt_124_daily['consumo_kwh'].interpolate(method='time')

print(f'Valores nulos após interpolação: {mt_124_daily.isnull().sum().sum()}')

display(mt_124_daily.describe())
```

```python
# Visualização da série temporal diária
fig, ax = plt.subplots(figsize=(16, 5))
mt_124_daily['consumo_kwh'].plot(ax=ax)

ax.set_title('Consumo diário de energia elétrica — Cliente MT_124')
ax.set_xlabel('Data')
ax.set_ylabel('Consumo diário em kWh')

plt.tight_layout()
plt.show()
```

```python
# Decomposição aditiva da série temporal
decomposicao = seasonal_decompose(
    mt_124_daily['consumo_kwh'],
    model='additive',
    period=365
)

fig = decomposicao.plot()
fig.set_size_inches(14, 10)
fig.suptitle('Decomposição aditiva — Consumo diário MT_124', fontsize=14, y=1.02)

plt.tight_layout()
plt.show()
```

```python
# Criação de variáveis sazonais
mt_124_daily['dia_semana'] = mt_124_daily.index.dayofweek
mt_124_daily['mes'] = mt_124_daily.index.month
mt_124_daily['dia_do_ano'] = mt_124_daily.index.dayofyear

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

sns.boxplot(
    data=mt_124_daily,
    x='dia_semana',
    y='consumo_kwh',
    ax=axes[0]
)
axes[0].set_title('Consumo por dia da semana')
axes[0].set_xlabel('Dia da semana: 0 = segunda, 6 = domingo')
axes[0].set_ylabel('Consumo em kWh')

sns.boxplot(
    data=mt_124_daily,
    x='mes',
    y='consumo_kwh',
    ax=axes[1]
)
axes[1].set_title('Consumo por mês')
axes[1].set_xlabel('Mês')
axes[1].set_ylabel('Consumo em kWh')

plt.tight_layout()
plt.show()
```

```python
# Análise de autocorrelação e autocorrelação parcial
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

plot_acf(mt_124_daily['consumo_kwh'], lags=60, ax=axes[0])
axes[0].set_title('Autocorrelação — ACF')

plot_pacf(mt_124_daily['consumo_kwh'], lags=60, ax=axes[1], method='ywm')
axes[1].set_title('Autocorrelação parcial — PACF')

plt.tight_layout()
plt.show()
```

```python
# Criação de atributos defasados e médias móveis
for lag in [1, 7, 30]:
    mt_124_daily[f'lag_{lag}'] = mt_124_daily['consumo_kwh'].shift(lag)

mt_124_daily['mm_7'] = mt_124_daily['consumo_kwh'].rolling(window=7).mean()
mt_124_daily['mm_30'] = mt_124_daily['consumo_kwh'].rolling(window=30).mean()

# Remoção dos valores ausentes gerados pelas defasagens
mt_124_daily.dropna(inplace=True)

print(f'Shape final após criação dos atributos: {mt_124_daily.shape}')
display(mt_124_daily.head())
```

```python
# Divisão temporal em treino e teste
treino = mt_124_daily.loc[:'2013-12-31']
teste = mt_124_daily.loc['2014-01-01':]

print(f'Treino: {treino.index.min().date()} a {treino.index.max().date()} — {len(treino)} amostras')
print(f'Teste: {teste.index.min().date()} a {teste.index.max().date()} — {len(teste)} amostras')

fig, ax = plt.subplots(figsize=(16, 5))

ax.plot(treino['consumo_kwh'], label='Treino')
ax.plot(teste['consumo_kwh'], label='Teste')

ax.set_title('Divisão treino/teste — Consumo diário MT_124')
ax.set_xlabel('Data')
ax.set_ylabel('Consumo em kWh')
ax.legend()

plt.tight_layout()
plt.show()
```

# **Modelos**

A etapa de modelagem foi conduzida com o objetivo de prever o consumo diário de energia elétrica do cliente **MT_124**, utilizando os dados históricos previamente tratados e agregados em frequência diária. A construção dos modelos seguiu uma abordagem progressiva, começando por métodos simples de referência e avançando para um modelo estatístico mais adequado à natureza sazonal da série temporal.

Inicialmente, foram definidos modelos baseline para estabelecer uma referência mínima de desempenho. O primeiro modelo utilizado foi o modelo de **persistência**, também conhecido como modelo naive. Nesse método, a previsão do consumo de um determinado dia é igual ao consumo observado no dia anterior. Apesar de sua simplicidade, esse modelo é importante em problemas de séries temporais, pois permite verificar se modelos mais complexos realmente conseguem superar uma estratégia básica baseada apenas no valor mais recente da série.

O segundo modelo utilizado foi a **média móvel de 7 dias**. Essa abordagem calcula a previsão a partir da média dos sete dias anteriores, buscando suavizar oscilações diárias e capturar parcialmente o comportamento semanal do consumo. A escolha da janela de 7 dias está relacionada aos padrões observados na análise exploratória, especialmente à diferença de comportamento entre dias úteis e finais de semana. Esse modelo tende a ser mais estável do que a persistência, embora possa apresentar dificuldade para acompanhar mudanças abruptas ou picos de consumo.

Após a construção dos modelos de referência, foi implementado o modelo **SARIMA**, uma extensão sazonal do ARIMA. A escolha desse modelo foi motivada pela presença de autocorrelação e sazonalidade semanal identificadas durante a etapa de EDA. O SARIMA permite representar componentes autorregressivos, diferenciação, médias móveis e padrões sazonais, sendo adequado para séries temporais com comportamento recorrente ao longo do tempo.

O modelo SARIMA foi configurado inicialmente como **SARIMA(1,1,1)(1,1,1,7)**. A primeira parte da configuração representa os componentes não sazonais do modelo, enquanto a segunda parte representa os componentes sazonais. O período sazonal igual a 7 foi adotado para capturar o ciclo semanal do consumo elétrico. A diferenciação foi utilizada para reduzir efeitos de tendência e tornar a série mais adequada à modelagem estatística.

Antes da aplicação do SARIMA, foi realizado o teste de Dickey-Fuller aumentado para avaliar a estacionariedade da série de treinamento. Esse teste verifica se a série apresenta raiz unitária, ou seja, se possui comportamento não estacionário. Como séries de consumo elétrico podem apresentar tendência e sazonalidade, essa etapa foi importante para justificar a aplicação de diferenciação no modelo.

A validação dos modelos foi realizada por meio de uma divisão temporal dos dados. Os anos de **2011 a 2013** foram utilizados como conjunto de treinamento, enquanto o ano de **2014** foi reservado para teste. Essa estratégia respeita a ordem natural da série temporal e evita vazamento de informação, pois os modelos são treinados apenas com dados passados e avaliados em um período futuro.

Para comparar o desempenho dos modelos, foram utilizadas as métricas **MAE**, **RMSE** e **MAPE**. O MAE mede o erro absoluto médio em kWh, permitindo interpretar o desvio médio das previsões em relação aos valores reais. O RMSE também mede o erro das previsões, mas penaliza mais fortemente erros maiores. Já o MAPE expressa o erro médio em termos percentuais, facilitando a interpretação do desempenho relativo do modelo.

O refinamento dos modelos ocorreu a partir dos padrões identificados na análise exploratória. A presença de sazonalidade semanal orientou tanto a escolha da média móvel de 7 dias quanto a configuração sazonal do SARIMA. Além disso, os gráficos de autocorrelação e autocorrelação parcial auxiliaram na identificação da dependência temporal da série, reforçando a necessidade de utilizar modelos capazes de considerar valores passados na previsão.

A comparação entre os modelos permite avaliar se abordagens mais complexas conseguem superar métodos simples de referência. O modelo de persistência funciona como um ponto de partida, pois considera apenas o consumo imediatamente anterior. A média móvel de 7 dias adiciona uma suavização semanal, sendo útil para reduzir oscilações de curto prazo. Já o SARIMA incorpora explicitamente componentes temporais e sazonais, sendo mais adequado para representar padrões recorrentes da série.

Dessa forma, a etapa de modelagem buscou equilibrar simplicidade, interpretabilidade e capacidade preditiva. Os modelos baseline foram importantes para estabelecer uma referência inicial, enquanto o SARIMA foi utilizado como uma alternativa estatística mais robusta para capturar a dinâmica temporal do consumo diário de energia elétrica. A escolha do melhor modelo deve considerar não apenas os menores valores de erro, mas também a estabilidade das previsões, a coerência com os padrões observados na EDA e a possibilidade de aplicação em cenários reais de previsão.

```python
#Modelos Propostos 

from sklearn.metrics import mean_absolute_error, mean_squared_error, mean_absolute_percentage_error
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.stattools import adfuller

# Função para cálculo das métricas
def calcular_metricas(y_true, y_pred, nome_modelo):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = mean_absolute_percentage_error(y_true, y_pred) * 100

    return {
        'Modelo': nome_modelo,
        'MAE': round(mae, 2),
        'RMSE': round(rmse, 2),
        'MAPE (%)': round(mape, 2)
    }

# Série real de teste
y_test = teste['consumo_kwh']

# Lista para armazenar os resultados
resultados = []
```

```python
# Modelo de persistência: previsão do dia atual igual ao consumo do dia anterior
y_test_shifted = teste['consumo_kwh'].shift(1)

y_pred_naive = y_test_shifted.copy()
y_pred_naive.iloc[0] = treino['consumo_kwh'].iloc[-1]

# Cálculo das métricas
resultados.append(
    calcular_metricas(y_test, y_pred_naive, 'Persistência (lag-1)')
)

# Visualização
fig, ax = plt.subplots(figsize=(16, 5))

ax.plot(teste.index, y_test, label='Observado')
ax.plot(teste.index, y_pred_naive, label='Persistência', alpha=0.8)

ax.set_title('Modelo de Persistência — Consumo diário MT_124')
ax.set_xlabel('Data')
ax.set_ylabel('Consumo em kWh')
ax.legend()

plt.tight_layout()
plt.show()
```

```python
# Junção da série de treino e teste para cálculo da média móvel
serie_completa = pd.concat([treino['consumo_kwh'], teste['consumo_kwh']])

# Previsão com média móvel dos 7 dias anteriores
y_pred_mm7 = serie_completa.shift(1).rolling(window=7).mean()
y_pred_mm7 = y_pred_mm7.loc[teste.index]
y_pred_mm7 = y_pred_mm7.dropna()

# Ajuste do y real para o mesmo índice da previsão
y_test_mm7 = y_test.loc[y_pred_mm7.index]

# Cálculo das métricas
resultados.append(
    calcular_metricas(y_test_mm7, y_pred_mm7, 'Média Móvel (7 dias)')
)

# Visualização
fig, ax = plt.subplots(figsize=(16, 5))

ax.plot(y_test.index, y_test, label='Observado')
ax.plot(y_pred_mm7.index, y_pred_mm7, label='Média Móvel 7 dias', alpha=0.8)

ax.set_title('Modelo de Média Móvel de 7 dias — Consumo diário MT_124')
ax.set_xlabel('Data')
ax.set_ylabel('Consumo em kWh')
ax.legend()

plt.tight_layout()
plt.show()
```

```python
# Teste de estacionariedade Dickey-Fuller aumentado
resultado_adf = adfuller(treino['consumo_kwh'])

print(f'Teste ADF — Estatística: {resultado_adf[0]:.4f}')
print(f'p-valor: {resultado_adf[1]:.4f}')

if resultado_adf[1] < 0.05:
    print('A série é estacionária, pois rejeita H0.')
else:
    print('A série não é estacionária, sendo necessário aplicar diferenciação.')
```

```python
# Ajuste do modelo SARIMA
modelo_sarima = SARIMAX(
    treino['consumo_kwh'],
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 7),
    enforce_stationarity=False,
    enforce_invertibility=False
)

resultado_sarima = modelo_sarima.fit(disp=False)

print(resultado_sarima.summary())
```

```python
# Previsão para o período de teste
y_pred_sarima = resultado_sarima.forecast(steps=len(teste))
y_pred_sarima.index = teste.index

# Cálculo das métricas
resultados.append(
    calcular_metricas(y_test, y_pred_sarima, 'SARIMA(1,1,1)(1,1,1,7)')
)

# Visualização
fig, ax = plt.subplots(figsize=(16, 5))

ax.plot(y_test.index, y_test, label='Observado')
ax.plot(y_test.index, y_pred_sarima, label='SARIMA', alpha=0.8)

ax.set_title('Modelo SARIMA(1,1,1)(1,1,1,7) — Consumo diário MT_124')
ax.set_xlabel('Data')
ax.set_ylabel('Consumo em kWh')
ax.legend()

plt.tight_layout()
plt.show()
```

```python
# Diagnóstico dos resíduos do modelo SARIMA
resultado_sarima.plot_diagnostics(figsize=(14, 8))
plt.tight_layout()
plt.show()
```

```python
# Visualização comparativa das métricas
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

metricas = ['MAE', 'RMSE', 'MAPE (%)']

for ax, metrica in zip(axes, metricas):
    df_resultados.plot(
        x='Modelo',
        y=metrica,
        kind='bar',
        ax=ax,
        legend=False,
        edgecolor='black'
    )

    ax.set_title(metrica)
    ax.set_ylabel(metrica)
    ax.tick_params(axis='x', rotation=15)

plt.suptitle('Comparação de métricas entre os modelos', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```
